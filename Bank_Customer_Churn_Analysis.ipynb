## Step 7: Unnecessary Columns Remove

## Step 8: Calculate Customer Churn KPIs

```python
total_customers = len(df_clean)
churn_customers = df_clean["Exited"].sum() #churn_cust mean exited
active_customers = total_customers - churn_customers

churn_rate = (churn_customers / total_customers) * 100
retention_rate = (active_customers / total_customers) * 100

print("Total Customers:", total_customers)
print("Exited Customers:", churn_customers)
print("Active Customers:", active_customers)
print("Churn Rate:", round(churn_rate, 2), "%")
print("Retention Rate:", round(retention_rate, 2), "%")
```

## Step 9: Geography-wise Churn Analysis

### Which country has the highest customer churn rate?

```python

geo_analysis = df_clean.groupby("Geography").agg(
    Total_Customers=("Exited", "count"),
    Exited_Customers=("Exited", "sum"),
    Churn_Rate=("Exited", "mean")
).reset_index()

geo_analysis["Churn_Rate"] *= 100

geo_analysis = geo_analysis.sort_values(
    by="Churn_Rate", ascending=False
)

display(geo_analysis.round(2))
```

```python

import plotly.express as px

fig = px.bar(
    geo_analysis,
    x="Geography",
    y="Churn_Rate",
    color="Geography",
    text_auto=".2f",
    title="Geography-wise Customer Churn Rate (%)"
)

fig.show()
```

## Step 10: Gender-wise Churn Analysis

### Which gender has a higher churn rate?


```python
gender_analysis = df_clean.groupby("Gender").agg(
    Total_Customers=("Exited", "count"),
    Exited_Customers=("Exited", "sum"),
    Churn_Rate=("Exited", "mean")
).reset_index()

gender_analysis["Churn_Rate"] *= 100

gender_analysis = gender_analysis.sort_values(
    by="Churn_Rate", ascending=False
)

display(gender_analysis.round(2))
```

## Step 11: Age Group Analysis

```python

df_clean["AgeGroup"] = pd.cut(
    df_clean["Age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "55+"]
)

age_analysis = df_clean.groupby(
    "AgeGroup", observed=False
).agg(
    Total_Customers=("Exited", "count"),
    Exited_Customers=("Exited", "sum"),
    Churn_Rate=("Exited", "mean")
).reset_index()

age_analysis["Churn_Rate"] *= 100

display(age_analysis.round(2))
```

## Step 12: Age Group Churn Chart

```python

fig = px.bar(
    age_analysis,
    x="AgeGroup",
    y="Churn_Rate",
    color="AgeGroup",
    title="Customer Churn Rate by Age Group",
    text_auto=".2f"
)

fig.show()
```

## Step 13: Product-wise Churn Analysis
### Business Question: Which product category has the highest customer churn rate?

```python

product_analysis = df_clean.groupby("NumOfProducts").agg(
    Total_Customers=("Exited", "count"),
    Exited_Customers=("Exited", "sum"),
    Churn_Rate=("Exited", "mean")
).reset_index()

product_analysis["Churn_Rate"] *= 100

display(product_analysis.round(2))
```

```python

fig = px.bar(
    product_analysis,
    x="NumOfProducts",
    y="Churn_Rate",
    color="NumOfProducts",
    text_auto=".2f",
    title="Churn Rate by Number of Products"
)

fig.show()
```

## Step 14: Active vs Inactive Members
### Business Question: Are inactive customers more likely to leave the bank?

```python

active_analysis = df_clean.groupby("IsActiveMember").agg(
    Total_Customers=("Exited", "count"),
    Exited_Customers=("Exited", "sum"),
    Churn_Rate=("Exited", "mean")
).reset_index()

active_analysis["Member_Status"] = active_analysis[
    "IsActiveMember"
].map({1: "Active", 0: "Inactive"})

active_analysis["Churn_Rate"] *= 100

display(active_analysis.round(2))
```

```python

fig = px.bar(
    active_analysis,
    x="Member_Status",
    y="Churn_Rate",
    color="Member_Status",
    text_auto=".2f",
    title="Active vs Inactive Customer Churn Rate"
)

fig.show()
```

## Step 15: Credit Card Analysis
### Business Question: Does having a credit card affect customer churn?

```python

credit_analysis = df_clean.groupby("HasCrCard").agg(
    Total_Customers=("Exited", "count"),
    Exited_Customers=("Exited", "sum"),
    Churn_Rate=("Exited", "mean")
).reset_index()

credit_analysis["Card_Status"] = credit_analysis[
    "HasCrCard"
].map({1: "Has Credit Card", 0: "No Credit Card"})

credit_analysis["Churn_Rate"] *= 100

display(credit_analysis.round(2))
```

```python

fig = px.bar(
    credit_analysis,
    x="Card_Status",
    y="Churn_Rate",
    color="Card_Status",
    text_auto=".2f",
    title="Credit Card vs Customer Churn Rate"
)

fig.show()
```

## Step 21: Create Interactive Filters

```python

import pandas as pd
import plotly.express as px
from IPython.display import display

# Create dashboard filters

geo_options = ["All"] + sorted(
    df_clean["Geography"].dropna().unique().tolist()
)

gender_options = ["All"] + sorted(
    df_clean["Gender"].dropna().unique().tolist()
)

selected_geo = "All"
selected_gender = "All"

# Filter function
def filter_data(data, geography, gender):

    filtered = data.copy()

    if geography != "All":
        filtered = filtered[
            filtered["Geography"] == geography
        ]

    if gender != "All":
        filtered = filtered[
            filtered["Gender"] == gender
        ]

    return filtered

print("Available Geography:", geo_options)
print("Available Gender:", gender_options)
```

## Step 22: Apply Filters

```python

# Select filters here
selected_geo = "France"
selected_gender = "Female"

filtered_df = filter_data(
    df_clean,
    selected_geo,
    selected_gender
)

print("Filtered Customers:", len(filtered_df))

display(filtered_df.head())
```

## Step 23: Dynamic KPI Cards

```python

# Calculate filtered KPIs

total = len(filtered_df)

exited = filtered_df["Exited"].sum()

retained = total - exited

churn_rate = (
    (exited / total) * 100 if total > 0 else 0
)

# Display KPI Cards

import plotly.graph_objects as go
from plotly.subplots import make_subplots

fig = make_subplots(
    rows=1,
    cols=4,
    specs=[[{"type": "indicator"}] * 4],
    subplot_titles=[
        "Total Customers",
        "Exited Customers",
        "Retained Customers",
        "Churn Rate"
    ]
)

values = [total, exited, retained, round(churn_rate, 2)]

for i, value in enumerate(values, start=1):
    fig.add_trace(
        go.Indicator(
            mode="number",
            value=value,
            number={"suffix": "%" if i == 4 else ""}
        ),
        row=1,
        col=i
    )

fig.update_layout(
    title="Bank Customer Churn KPIs",
    height=250
)

fig.show()
```

## Step 24: Multiple Charts in One Dashboard
आता Geography, Gender, Age आणि Product Analysis एकाच dashboard मध्ये दाखवू.

```python

# Prepare analysis data

geo_chart = filtered_df.groupby(
    "Geography"
)["Exited"].mean().mul(100).reset_index()

gender_chart = filtered_df.groupby(
    "Gender"
)["Exited"].mean().mul(100).reset_index()

age_chart = filtered_df.groupby(
    "AgeGroup",
    observed=False
)["Exited"].mean().mul(100).reset_index()

product_chart = filtered_df.groupby(
    "NumOfProducts"
)["Exited"].mean().mul(100).reset_index()


# Create 2x2 Dashboard

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=[
        "Geography-wise Churn",
        "Gender-wise Churn",
        "Age Group Churn",
        "Product-wise Churn"
    ]
)

fig.add_trace(
    go.Bar(
        x=geo_chart["Geography"],
        y=geo_chart["Exited"],
        name="Geography"
    ),
    row=1, col=1
)

fig.add_trace(
    go.Bar(
        x=gender_chart["Gender"],
        y=gender_chart["Exited"],
        name="Gender"
    ),
    row=1, col=2
)

fig.add_trace(
    go.Bar(
        x=age_chart["AgeGroup"].astype(str),
        y=age_chart["Exited"],
        name="Age Group"
    ),
    row=2, col=1
)

fig.add_trace(
    go.Bar(
        x=product_chart["NumOfProducts"],
        y=product_chart["Exited"],
        name="Products"
    ),
    row=2, col=2
)

fig.update_layout(
    title="Customer Churn Analysis Dashboard",
    height=700,
    showlegend=False
)

fig.update_yaxes(title_text="Churn Rate (%)")

fig.show()
```

## Step 25: Add Interactive Dropdown Filters
आता आपण dashboard मध्ये actual dropdown filters तयार करणार आहोत. User Geography आणि Gender select करू शकेल आणि सर्व KPIs आणि charts automatically update होतील.

```python

import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# Dropdown filters

geo_dropdown = widgets.Dropdown(
    options=["All", "France", "Germany", "Spain"],
    value="All",
    description="Geography:"
)

gender_dropdown = widgets.Dropdown(
    options=["All", "Female", "Male"],
    value="All",
    description="Gender:"
)

output = widgets.Output()

# Dashboard function

def update_dashboard(geography, gender):

    with output:
        clear_output(wait=True)

        data = filter_data(
            df_clean,
            geography,
            gender
        )

        total = len(data)
        exited = data["Exited"].sum()
        retained = total - exited

        rate = (
            exited / total * 100 if total > 0 else 0
        )

        # KPI cards

        display(HTML(f"""
        <div style="display:flex; gap:12px; flex-wrap:wrap;">
          <div style="background:#e8f1ff;padding:18px;border-radius:10px;flex:1">
            <b>Total Customers</b><h2>{total:,}</h2>
          </div>
          <div style="background:#ffe8e8;padding:18px;border-radius:10px;flex:1">
            <b>Exited Customers</b><h2>{exited:,}</h2>
          </div>
          <div style="background:#e6f7ed;padding:18px;border-radius:10px;flex:1">
            <b>Retained Customers</b><h2>{retained:,}</h2>
          </div>
          <div style="background:#fff3d9;padding:18px;border-radius:10px;flex:1">
            <b>Churn Rate</b><h2>{rate:.2f}%</h2>
          </div>
        </div>
        """))

        # Charts

        charts = [
            ("Geography", "Geography-wise Churn"),
            ("Gender", "Gender-wise Churn"),
            ("AgeGroup", "Age-wise Churn"),
            ("NumOfProducts", "Product-wise Churn")
        ]

        for column, title in charts:

            result = data.groupby(
                column, observed=False
            )["Exited"].mean().mul(100).reset_index()

            fig = px.bar(
                result,
                x=column,
                y="Exited",
                title=title,
                labels={"Exited": "Churn Rate (%)"},
                text_auto=".2f"
            )

            fig.update_layout(height=350)
            fig.show()

        # Customer table

        display(HTML("<h3>Customer Details</h3>"))

        display(data.head(20))


# Connect filters

controls = widgets.VBox([
    geo_dropdown,
    gender_dropdown
])

dashboard = widgets.interactive_output(
    update_dashboard,
    {
        "geography": geo_dropdown,
        "gender": gender_dropdown
    }
)

display(controls, output)
```

## Step 27: Customer Risk Analysis

```python

risk_df = df_clean.copy()

def customer_risk(row):

    if row["IsActiveMember"] == 0 and row["NumOfProducts"] == 1:
        return "High Attention"

    elif row["IsActiveMember"] == 0:
        return "Medium Attention"

    else:
        return "Regular"

risk_df["CustomerSegment"] = risk_df.apply(
    customer_risk,
    axis=1
)

display(
    risk_df[
        ["Geography", "Gender", "Age",
         "IsActiveMember", "NumOfProducts",
         "Exited", "CustomerSegment"]
    ].head(10)
)
```

## Step 28: Customer Segment Analysis

```python

segment_analysis = risk_df.groupby(
    "CustomerSegment"
).agg(
    Total_Customers=("Exited", "count"),
    Exited_Customers=("Exited", "sum"),
    Churn_Rate=("Exited", "mean")
).reset_index()

segment_analysis["Churn_Rate"] *= 100

display(segment_analysis.round(2))
```

```python

fig = px.bar(
    segment_analysis,
    x="CustomerSegment",
    y="Churn_Rate",
    color="CustomerSegment",
    text_auto=".2f",
    title="Customer Segment-wise Churn Rate"
)

fig.show()
```

```python

%%writefile app.py

import streamlit as st
import pandas as pd
import plotly.express as px
import os

# Page Configuration
st.set_page_config(
    page_title="Bank Churn Dashboard",
    page_icon="🏦",
    layout="wide"
)

st.title("🏦 Bank Customer Churn Dashboard")
st.caption("Interactive Customer Retention & Churn Analysis")

# Load Dataset
uploaded_file = st.sidebar.file_uploader(
    "Upload Bank Dataset",
    type=["csv"]
)

if uploaded_file is not None:
    df = pd.read_csv(uploaded_file)
elif os.path.exists("Churn_Modelling.csv"):
    df = pd.read_csv("Churn_Modelling.csv")
else:
    st.info("Please upload Churn_Modelling.csv")
    st.stop()

# Data Preparation
df.columns = df.columns.str.strip()
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]

df = df.drop(
    columns=["RowNumber", "CustomerId", "Surname"],
    errors="ignore"
)

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "55+"]
)

# Sidebar Filters
st.sidebar.header("Dashboard Filters")

geographies = ["All"] + sorted(
    df["Geography"].dropna().unique().tolist()
)

genders = ["All"] + sorted(
    df["Gender"].dropna().unique().tolist()
)

selected_geo = st.sidebar.selectbox(
    "Select Geography", geographies
)

selected_gender = st.sidebar.selectbox(
    "Select Gender", genders
)

filtered = df.copy()

if selected_geo != "All":
    filtered = filtered[
        filtered["Geography"] == selected_geo
    ]

if selected_gender != "All":
    filtered = filtered[
        filtered["Gender"] == selected_gender
    ]

# KPI Calculations
total = len(filtered)
exited = int(filtered["Exited"].sum())
retained = total - exited

churn_rate = (
    exited / total * 100 if total else 0
)

# KPI Cards
st.subheader("Business KPIs")

c1, c2, c3, c4 = st.columns(4)

c1.metric("Total Customers", f"{total:,}")
c2.metric("Exited Customers", f"{exited:,}")
c3.metric("Retained Customers", f"{retained:,}")
c4.metric("Churn Rate", f"{churn_rate:.2f}%")

st.divider()

# Geography and Gender Charts
st.subheader("Customer Churn Analysis")

col1, col2 = st.columns(2)

with col1:
    geo = filtered.groupby(
        "Geography"
    )["Exited"].mean().mul(100).reset_index()

    fig = px.bar(
        geo,
        x="Geography",
        y="Exited",
        color="Geography",
        title="Geography-wise Churn Rate",
        labels={"Exited": "Churn Rate (%)"}
    )

    st.plotly_chart(fig, use_container_width=True)

with col2:
    gender = filtered.groupby(
        "Gender"
    )["Exited"].mean().mul(100).reset_index()

    fig = px.pie(
        gender,
        names="Gender",
        values="Exited",
        title="Gender-wise Churn Rate",
        hole=0.4
    )

    st.plotly_chart(fig, use_container_width=True)

# Age and Product Analysis
col3, col4 = st.columns(2)

with col3:
    age = filtered.groupby(
        "AgeGroup",
        observed=False
    )["Exited"].mean().mul(100).reset_index()

    fig = px.bar(
        age,
        x="AgeGroup",
        y="Exited",
        title="Age-wise Churn Rate",
        labels={"Exited": "Churn Rate (%)"}
    )

    st.plotly_chart(fig, use_container_width=True)

with col4:
    product = filtered.groupby(
        "NumOfProducts"
    )["Exited"].mean().mul(100).reset_index()

    fig = px.bar(
        product,
        x="NumOfProducts",
        y="Exited",
        title="Product-wise Churn Rate",
        labels={"Exited": "Churn Rate (%)"}
    )

    st.plotly_chart(fig, use_container_width=True)

# Active Member Analysis
st.subheader("Active Member Analysis")

active = filtered.groupby(
    "IsActiveMember"
)["Exited"].mean().mul(100).reset_index()

active["Member Status"] = active[
    "IsActiveMember"
].map({1: "Active", 0: "Inactive"})

fig = px.bar(
    active,
    x="Member Status",
    y="Exited",
    color="Member Status",
    title="Active vs Inactive Churn Rate",
    labels={"Exited": "Churn Rate (%)"}
)

st.plotly_chart(fig, use_container_width=True)

# Customer Data
st.subheader("Customer Details")

status = st.selectbox(
    "Customer Status",
    ["All", "Exited", "Retained"]
)

if status == "Exited":
    display_data = filtered[filtered["Exited"] == 1]
elif status == "Retained":
    display_data = filtered[filtered["Exited"] == 0]
else:
    display_data = filtered

st.dataframe(
    display_data,
    use_container_width=True,
    hide_index=True
)

# Download Filtered Data
csv = display_data.to_csv(index=False).encode("utf-8")

st.download_button(
    "Download Customer Data",
    data=csv,
    file_name="filtered_customers.csv",
    mime="text/csv"
)

st.caption("Bank Customer Churn Analysis | Python & Streamlit")
```
